```python
import re
import pandas as pd
```

```python
with open("/home/jhin/program/trabalhos_projeto_e_analise_de_algoritmos_DIM0605/data/data/j30/j3010_1.sm", "r") as file:
    lines = file.readlines()
```

```python
sections = {
    "PROJECT_INFORMATION": [],
    "PRECEDENCE_RELATIONS": [],
    "REQUESTS/DURATIONS": [],
    "RESOURCEAVAILABILITIES": [],
}
current_section = None
```

```python
for line in lines:
    line = line.strip()

    if line.startswith("PROJECT INFORMATION"):
        current_section = "PROJECT_INFORMATION"
        continue
    elif line.startswith("PRECEDENCE RELATIONS"):
        current_section = "PRECEDENCE_RELATIONS"
        continue
    elif line.startswith("REQUESTS/DURATIONS"):
        current_section = "REQUESTS/DURATIONS"
        continue
    elif line.startswith("RESOURCEAVAILABILITIES"):
        current_section = "RESOURCEAVAILABILITIES"
        continue

    if current_section and line:
        sections[current_section].append(line)
```

```python
project_info = {}
for line in sections["PROJECT_INFORMATION"][1:]:
    if not line.startswith("*"):
        parts = re.split(r"\s+", line)
        if len(parts) >= 6:
            project_info = {
                "jobs": int(parts[1]),
                "release_date": int(parts[2]),
                "due_date": int(parts[3]),
                "tard_cost": int(parts[4]),
                "mpm_time": int(parts[5]),
            }
```

```python
project_info
```

```python
relations = []
for line in sections["PRECEDENCE_RELATIONS"][1:]:
    if not line.startswith("*"):
        parts = re.split(r"\s+", line)
        jobnr = int(parts[0])
        modes = int(parts[1])
        successorsCounter = int(parts[2])
        successors = list(map(int, parts[3:]))
        relations.append(
            {
                "jobnr": jobnr,
                "modes": modes,
                "successorsCounter": successorsCounter,
                "successors": successors,
            }
        )
    #return pd.DataFrame(relations)
```

```python
relations
```

```python
durations = []
for line in sections["REQUESTS/DURATIONS"][2:]:
    if not line.startswith("*"):
        parts = re.split(r"\s+", line)
        if len(parts) >= 6:
            jobnr = int(parts[0])
            mode = int(parts[1])
            duration = int(parts[2])
            resources = list(map(int, parts[3:]))
            durations.append(
                {
                    "jobnr": jobnr,
                    "mode": mode,
                    "duration": duration,
                    "resources": resources,
                }
            )
```

```python
durations
```

```python
resources = {}
for line in sections["RESOURCEAVAILABILITIES"][1:]:
    if not line.startswith("*"):
        parts = re.split(r"\s+", line)
        if len(parts) >= 4:
            r1 = int(parts[0])
            r2 = int(parts[1])
            r3 = int(parts[2])
            r4 = int(parts[3])
            resources = {
                "R1": r1,
                "R2": r2,
                "R3": r3,
                "R4": r4,
            }
```
